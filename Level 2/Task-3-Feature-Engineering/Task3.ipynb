
## Level 2 - Task 3: Feature Engineering

### Objective

Create new features from the restaurant dataset to improve analysis and
prepare the data for further machine learning tasks.


### Import libraries

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("Cleaned_Dataset.csv")

print("Dataset shape:", df.shape)
display(df.head())
```

```python
print("Original columns:")
print(df.columns.tolist())

print("\nMissing values:")
display(df.isnull().sum())
```

```python
df["Restaurant Name Length"] = df["Restaurant Name"].str.len()

display(
    df[["Restaurant Name", "Restaurant Name Length"]].head(10)
)
```

```python
df["Address Length"] = df["Address"].str.len()

display(
    df[["Address", "Address Length"]].head(10)
)
```

```python
df["Has Table Booking Encoded"] = df["Has Table booking"].map({
    "Yes": 1,
    "No": 0
})

display(
    df[["Has Table booking", "Has Table Booking Encoded"]].head(10)
)
```

```python
df["Has Online Delivery Encoded"] = df["Has Online delivery"].map({
    "Yes": 1,
    "No": 0
})

display(
    df[["Has Online delivery", "Has Online Delivery Encoded"]].head(10)
)
```

```python
feature_columns = [
    "Restaurant Name Length",
    "Address Length",
    "Has Table Booking Encoded",
    "Has Online Delivery Encoded"
]

display(df[feature_columns].head(10))
```

```python
print("Updated dataset shape:", df.shape)

print("\nNew columns:")
print(feature_columns)

print("\nMissing values in new features:")
display(df[feature_columns].isnull().sum())
```

```python
df.to_csv("Feature_Engineered_Dataset.csv", index=False)

print("Feature-engineered dataset saved successfully!")
print("Final shape:", df.shape)
```

```python
print("Feature Engineering Summary")

print("\nFeature columns:")
for column in feature_columns:
    print("-", column)

print("\nData types:")
display(df[feature_columns].dtypes)
```

```python
print("Statistical summary of engineered features:")

display(df[feature_columns].describe())
```

```python
print("Final feature-engineered dataset:")

display(
    df[
        [
            "Restaurant Name",
            "Address",
            "Restaurant Name Length",
            "Address Length",
            "Has Table Booking Encoded",
            "Has Online Delivery Encoded"
        ]
    ].head(10)
)
```

```python
print("Final dataset shape:", df.shape)

print("\nTotal missing values:", df.isnull().sum().sum())

print("Total duplicate rows:", df.duplicated().sum())
```

```python
print("Final columns:")
for i, column in enumerate(df.columns, start=1):
    print(i, "-", column)
```

```python
output_file = "Feature_Engineered_Dataset.csv"

df.to_csv(output_file, index=False)

print(f"Dataset saved successfully as: {output_file}")
```
